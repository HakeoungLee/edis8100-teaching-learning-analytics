## The five questions

van Leeuwen, Teasley, and Wise (2022) lay out a design space for teacher-facing and student-facing analytics. Five of its coordinates do most of the work in a critique, so these are the five we ask of every screen below.

| | |
|---|---|
| **1. Who is it for?** | A teacher, a student, both, or somebody else such as an advisor or an administrator. And the second half of the question: who else can see it. |
| **2. What does it compare you to?** | Your own past, a stated goal, the class, an expert model. This is the reference frame, and it decides most of what a display means. "there is the question of what reference point the data should be compared to (e.g. a pre-determined standard or relative values for peers)" (p. 131). |
| **3. How much interpreting has it already done?** | **Mirroring** reflects activity back with no judgment. **Alerting** flags a departure from an expected state. **Advising** recommends an action. Each step moves interpretive work out of the person and into the system, and moves responsibility with it. |
| **4. What can the viewer do about it?** | Agency. In cognitive tutors "the computer is in control, whereas dashboards offer information to students, who decide on any possible follow-up actions" (p. 133). |
| **5. Where does it sit?** | Extracted analytics show traces "in an interface separate from the learning environment that generated them"; embedded analytics show them "directly in the learning environment that generated them" (p. 132). Pushed to you, or somewhere you have to go and look. |

Hold one more question underneath all five, from Wise and Jung (2019): **what would a teacher or student do differently on Monday because of this screen, and where would that action go wrong?**

## Dashboard 1. Check My Activity

**What it was.** One of the first wide-scale student-facing dashboards, deployed at the University of Maryland, Baltimore County. It let university students compare their own LMS activity and their grades against their classmates. van Leeuwen, Teasley, and Wise (2022) cite it as an early large deployment of the student-facing idea.

**What was found.** Student attention went to the grade views. The chapter notes that "student focus on grade views has been observed repeatedly."

**Read it with the five questions.** Take two minutes on these before anyone says whether it is good.

1. Who is it for, and who else can see it?
2. What does it compare you to? What does choosing *the class* as the comparison presume about the student reading it?
3. Mirroring, alerting, or advising?
4. A student sees they are below the class on activity. What can they do at that moment?
5. Does it sit inside the course site, or somewhere separate?

**Then the one that matters.** If student attention lands on the grade rather than the activity, is that a failure of the student or a property of the design? What on the screen produced it?

## Dashboard 2. The Blackboard-derived dashboard

**What it was.** Corrin and de Barba built a dashboard following Blackboard's design, showing students their formative and summative assessment scores alongside their LMS engagement. Cited in van Leeuwen, Teasley, and Wise (2022).

**What was found.** This is the sentence to sit with. Students *were* able to articulate and interpret the feedback the dashboard presented. But there was "little evidence of students' ability to understand the connection between feedback and their current learning strategies."

So the display worked and the loop did not close. They could read it. They could not get from reading it to changing what they do.

**Read it with the five questions**, and then:

1. Which of the five questions does that finding land on? Argue for one.
2. The dashboard shows scores and engagement side by side. What does putting those two things on one screen imply about their relationship, and does the data support that implication?
3. What would have to be added, removed, or relabeled before a student could get from this screen to a different study session? Be specific about what appears and where.

## Dashboard 3. The weekly message

**What it was.** A message-based analytics tool in the OnTask tradition. Pardo and colleagues (2019) built OnTask to send students weekly automated personalized emails summarizing their progress on engagement metrics with recommendations and links to resources. Jung and Wise (2025) studied a version of this in an online undergraduate statistics course, where 91 students received weekly messages about a collaborative annotation task over five weeks.

**What the message looked like.** Open Jung and Wise (2025) and look at the message figure in the paper, or at the slide. It greets the student, says what is happening in the course reading this week and when it is due, and then shows an "active conversation" from the class: a highlighted passage with two of the comments other students left on it, and a **View conversation** button that drops the student straight into that spot in the reading.

**What was found.** Three things, and each is a different kind of finding.

- **Access was high and stayed high.** 77 percent of students opened the analytics in the first week, and between 81 and 95 percent over the four weeks after.
- **Clicking was not the same as acting.** Students did not always act immediately and directly on what they were shown. What the researchers found instead were indirect changes: heightened awareness, and behavioral adjustments during later collaborative tasks.
- **Students wanted two things that pull apart.** They wanted concise data, and they wanted enough context to interpret and evaluate it. Those two requests are in tension, and a design has to choose.

For contrast, the same paper reports a dashboard study of 274 chemistry undergraduates across three semesters: overall access ran between 55 and 73 percent, but the students who came back with any frequency were only 13 to 24 percent.

**Read it with the five questions**, and then:

1. This one is pushed to the student rather than waiting to be visited, and it is embedded in the task rather than separate from it. Which of the three tools got the most engagement, and which design choices would you credit?
2. High access and low direct action. What does that pair mean for how you would measure whether your own dashboard worked?
3. Concise against enough context. Where would you put that dial, and who does your choice disadvantage?
